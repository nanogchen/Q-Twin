```python
import numpy as np
import matplotlib.pyplot as plt
import MDAnalysis as mda
import sys,os
sys.path.insert(0, "../../..")
from srcs.calc import get_scattering_image
from srcs.q_gen import get_binning_averages
```

```python
u = mda.Universe("../../../data/colloids/L30_vf60_rate1.0.data") 
system = u.select_atoms('all')
bx, by, bz = u.dimensions[:3]
box = np.array([bx, by, bz])
```

```python
q_max = 10
# note in the paper q_max=20 was used. 
```

```python
len(u.trajectory)
```

```python
system.atoms.n_atoms
```

```python
box
```

```python
%%time
# scattering in xz
_,_,qpts1, qpts2, ssf_reshape = get_scattering_image(box, q_max, system, u.trajectory[0], plane='xz')
ssf_reshape = np.squeeze(ssf_reshape)
# rm the center pt
ssf_reshape[int(ssf_reshape.shape[0]/2), int(ssf_reshape.shape[1]/2)] = 0.0

plt.clf()
plt.contourf(qpts1, qpts2, ssf_reshape.transpose(), cmap='jet')
plt.colorbar()
plt.axis('equal')
plt.xlabel(r"$q_x\,(\sigma^{-1})$")
plt.ylabel(r"$q_z\,(\sigma^{-1})$")
plt.show()
```

```python
%%time
# scattering in xy
_,_,qpts1, qpts2, ssf_reshape = get_scattering_image(box, q_max, system, u.trajectory[0], plane='xy')
ssf_reshape = np.squeeze(ssf_reshape)
# rm the center pt
ssf_reshape[int(ssf_reshape.shape[0]/2), int(ssf_reshape.shape[1]/2)] = 0.0

plt.clf()
plt.contourf(qpts1, qpts2, ssf_reshape.transpose(), cmap='jet')
plt.colorbar()
plt.axis('equal')
plt.xlabel(r"$q_x\,(\sigma^{-1})$")
plt.ylabel(r"$q_y\,(\sigma^{-1})$")
plt.show()
```
